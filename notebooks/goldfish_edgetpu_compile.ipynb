# Goldfish Model - EdgeTPU Compilation
## Convert TFLite to EdgeTPU Format

This notebook takes your `goldfish_best_int8.tflite` and compiles it for the Coral EdgeTPU.

**Instructions:**
1. Upload your `goldfish_best_int8.tflite` file
2. Run all cells
3. Download the `goldfish_best_edgetpu.tflite` file

**Time:** ~2 minutes

## Step 1: Install EdgeTPU Compiler

```python
print("📦 Installing EdgeTPU compiler...\n")
!curl https://packages.cloud.google.com/apt/doc/apt-key.gpg | sudo apt-key add -
!echo "deb https://packages.cloud.google.com/apt coral-edgetpu-stable main" | sudo tee /etc/apt/sources.list.d/coral-edgetpu.list
!sudo apt-get update -qq
!sudo apt-get install edgetpu-compiler -y -qq

print("\n✅ EdgeTPU compiler installed!")
!edgetpu_compiler --version
```

## Step 2: Upload TFLite Model

Upload your `goldfish_best_int8.tflite` file.

```python
from google.colab import files
import os

print("📤 Upload your goldfish_best_int8.tflite file:")
print("(Click 'Choose Files' below)\n")

uploaded = files.upload()

if 'goldfish_best_int8.tflite' in uploaded:
    size_mb = len(uploaded['goldfish_best_int8.tflite']) / (1024*1024)
    print(f"\n✅ Uploaded goldfish_best_int8.tflite ({size_mb:.1f} MB)")
else:
    print("\n⚠️ Please upload goldfish_best_int8.tflite")
    print(f"You uploaded: {list(uploaded.keys())}")
```

## Step 3: Compile for EdgeTPU

```python
print("="*60)
print("COMPILING FOR EDGETPU")
print("="*60)

print("\n🔄 Compiling goldfish_best_int8.tflite...\n")

!edgetpu_compiler goldfish_best_int8.tflite

print("\n" + "="*60)

# Check if compilation succeeded
if os.path.exists('goldfish_best_int8_edgetpu.tflite'):
    size_mb = os.path.getsize('goldfish_best_int8_edgetpu.tflite') / (1024*1024)
    print(f"✅ EdgeTPU model created!")
    print(f"   File: goldfish_best_int8_edgetpu.tflite")
    print(f"   Size: {size_mb:.2f} MB")
else:
    print("❌ Compilation failed!")
    print("Check the output above for errors.")
```

## Step 4: Download EdgeTPU Model

```python
import shutil

# Rename to simpler filename
if os.path.exists('goldfish_best_int8_edgetpu.tflite'):
    shutil.copy('goldfish_best_int8_edgetpu.tflite', 'goldfish_best_edgetpu.tflite')
    
    print("📥 Downloading goldfish_best_edgetpu.tflite...\n")
    files.download('goldfish_best_edgetpu.tflite')
    
    print("\n✅ Download complete!")
    print("\nNext steps:")
    print("1. The file is now in your Downloads folder")
    print("2. Upload it to your Raspberry Pi")
    print("3. Test goldfish detection with the EdgeTPU!")
    print("\nExpected performance: 20-30 FPS on Coral TPU")
else:
    print("❌ EdgeTPU model not found!")
    print("Please check the compilation step above.")
```
