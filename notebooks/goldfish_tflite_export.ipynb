# Goldfish Model - TFLite Export
## Simple Export of Trained Model to TFLite Format

This notebook takes your trained `goldfish_best.pt` model and exports it to TFLite INT8 format for EdgeTPU.

**Instructions:**
1. Upload your `goldfish_best.pt` file (from Google Drive download)
2. Run all cells
3. Download the `goldfish_best_int8.tflite` file

**Time:** ~5 minutes

## Step 1: Install YOLOv8

```python
print("📦 Installing YOLOv8...")
!pip install ultralytics -q

from ultralytics import YOLO
import ultralytics
print(f"✅ Ultralytics version: {ultralytics.__version__}")
```

## Step 2: Upload Your Trained Model

Click the upload button below and select `goldfish_best.pt` from your computer.

```python
from google.colab import files
import os

print("📤 Upload your goldfish_best.pt file:")
print("(Click 'Choose Files' below)\n")

uploaded = files.upload()

# Check what was uploaded
if 'goldfish_best.pt' in uploaded:
    size_mb = len(uploaded['goldfish_best.pt']) / (1024*1024)
    print(f"\n✅ Uploaded goldfish_best.pt ({size_mb:.1f} MB)")
else:
    print("\n❌ Please upload goldfish_best.pt")
    print(f"You uploaded: {list(uploaded.keys())}")
```

## Step 3: Export to TFLite INT8

This converts your PyTorch model to TFLite format with INT8 quantization (required for EdgeTPU).

```python
import time

print("="*60)
print("EXPORTING TO TFLITE INT8 FORMAT")
print("="*60)

# Load the model
print("\n📦 Loading goldfish_best.pt...")
model = YOLO('goldfish_best.pt')
print("✅ Model loaded!")

# Export to TFLite
print("\n🔄 Exporting to TFLite INT8...")
print("This will take 3-5 minutes...\n")

start = time.time()

model.export(
    format='tflite',
    imgsz=640,
    int8=True,
)

elapsed = (time.time() - start) / 60
print(f"\n✅ Export complete in {elapsed:.1f} minutes!")
```

## Step 4: Find and Download TFLite File

```python
import glob
import shutil

print("🔍 Locating TFLite file...\n")

# Find the TFLite file
tflite_files = glob.glob('*_saved_model/*_int8.tflite')
if not tflite_files:
    tflite_files = glob.glob('*_saved_model/*.tflite')
if not tflite_files:
    tflite_files = glob.glob('*.tflite')

if tflite_files:
    tflite_file = tflite_files[0]
    print(f"✅ Found: {tflite_file}")
    
    # Copy to a simple filename
    output_name = 'goldfish_best_int8.tflite'
    shutil.copy(tflite_file, output_name)
    
    size_mb = os.path.getsize(output_name) / (1024*1024)
    print(f"Size: {size_mb:.2f} MB")
    print(f"\n📥 Downloading {output_name}...")
    
    files.download(output_name)
    
    print("\n✅ Download complete!")
    print("\nNext steps:")
    print("1. The file is now in your Downloads folder")
    print("2. We'll upload it to your Raspberry Pi")
    print("3. Convert it to EdgeTPU format on the Pi")
else:
    print("❌ TFLite file not found!")
    print("\nSearching for any .tflite files:")
    !find . -name "*.tflite" -o -name "*saved_model*"
    print("\nPlease share this output so we can troubleshoot.")
```
